# Conditional Statement
| Statement / Syntax                      | Short Description                                        |
| --------------------------------------- | -------------------------------------------------------- |
| `if`                                    | Executes a block of code when the condition is **True**. |
| `if-else`                               | Chooses between two blocks depending on **True/False**.  |
| `if-elif-else`                          | Checks multiple conditions in sequence.                  |
| Nested `if-else`                             | An `if` statement inside another `if` block.        |
| Ternary expression (`a if cond else b`) | Short, one-line conditional expression.                  |


```python
#if else 
num = 8

if num % 2 == 0:
    print("Number is even")
else:
    print("Number is odd")



#if elif else
marks = 85
if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 40:
    print("Grade D")
else:
    print("Fail")

#nested if else
age = 25
citizen = True

if age >= 18:
    if citizen == True:
        print("Eligible to vote")
    else:
        print("You must be a citizen")
else:
    print("You are under 18")

#ternary expression
age = 17
result = "Adult" if age >= 18 else "Minor"
print(result)
```

# Loops and Control Statements
| Loop / Statement | Short Description                                                       |
| ---------------- | -------------------------------------------------                       |
| `for` loop       | Loops over a sequence (list, range, string, etc.) - For Know Iterations |

```python
for i in range(1, 6):
    print(i)

fruits = ["apple", "banana", "mango"]
for fruit in fruits:
    print(fruit)

num = 5
for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

```python
i = 1
total = 0

while i <= 10:
    total = total + i
    i += 1

print("Total:", total)
```

```python
#continue: to skip the iteration
for i in range(1, 6):
    if i == 3:
        continue
    print(i)

#break: exits loops immediately
for i in range(1, 6):
    if i == 3:
        break
    print(i)
```

## List
A List in Python is an ordered, mutable collection used to store multiple items in a single variable.
It can contain elements of different data types and supports indexing, slicing, and modification.

| Method             | Description                                                 |
| ------------------ | ----------------------------------------------------------- |
| `append(x)`        | Adds item `x` to the end of the list.                       |
| `extend(iterable)` | Adds all elements of an iterable to the end.                |
| `insert(i, x)`     | Inserts item `x` at index `i`.                              |
| `remove(x)`        | Removes the first occurrence of item `x`.                   |
| `pop(i)`           | Removes and returns item at index `i` (default: last item). |
| `clear()`          | Removes all items from the list.                            |
| `index(x)`         | Returns the index of the first occurrence of `x`.           |
| `count(x)`         | Returns how many times `x` appears in the list.             |
| `sort()`           | Sorts the list in ascending order (in-place).               |
| `reverse()`        | Reverses the list (in-place).                               |
| `copy()`           | Returns a shallow copy of the list.                         |


```python
lst1= [1, 2, 3, 6, 4, 3, 1]
lst1.append(3)
print(lst1)
print(lst1.index(3))
print(lst1)
print(lst1.count(3))
lst1.sort()
print(lst1)
lst1.reverse()
print(lst1)
lst1=["apple", "banana", "mango"]
lst2=["cherry", "pineapple"]
lst1.extend(lst2)
print(lst1)
lst1.insert(1,"guava")
print(lst1)
lst1.remove("apple")
print(lst1)
lst1.pop(3)
print(lst1)
lst1.pop()
print(lst1)
lst1.clear()
print(lst1)
lst3=["apple", "banana", "mango"]
lst4= lst3.copy()
print(lst4)
```

Once created, you cannot change, add, or remove items.

```python
tup= (1,3,4,6,3,2,9)
print(len(tup))
print(max(tup))
print(min(tup))
print(sum(tup))
print(sorted(tup))
print(tuple([1,2,3]))
```

## 6.3 Sets
A set in Python is an unordered, mutable collection of unique elements.
It does not allow duplicates and does not maintain order.

| Method                              | Description                               | Application                      | Example (Input → Output)                                  |
| ----------------------------------- | ----------------------------------------- | -------------------------------- | --------------------------------------------------------- |
| **add(x)**                          | Adds element `x` to the set               | Insert new item                  | `{1,2}.add(3)` → `{1,2,3}`                                |
| **clear()**                         | Removes all elements                      | Reset set                        | `{1,2}.clear()` → `set()`                                 |
| **copy()**                          | Returns shallow copy                      | Duplicate set                    | `s.copy()` → `{…}`                                        |
| **difference(s2)**                  | Elements in s1 but not in s2              | Find non-overlapping items       | `{1,2,3}.difference({2,3})` → `{1}`                       |
| **difference_update(s2)**           | Removes elements of s2 from original set  | In-place difference              | `s={1,2,3}; s.difference_update({2}); s` → `{1,3}`        |
| **discard(x)**                      | Removes `x` if exists (no error)          | Safe removal                     | `{1,2}.discard(2)` → `{1}`                                |
| **intersection(s2)**                | Elements common in both sets              | Find shared items                | `{1,2,3}.intersection({2,3,4})` → `{2,3}`                 |
| **intersection_update(s2)**         | Keeps only common elements (in-place)     | In-place intersection            | `s={1,2,3}; s.intersection_update({2,3})` → `{2,3}`       |
| **isdisjoint(s2)**                  | True if no common elements                | Check if two sets overlap        | `{1,2}.isdisjoint({3})` → `True`                          |
| **issubset(s2)**                    | Checks if set is subset of s2             | Check containment                | `{1,2}.issubset({1,2,3})` → `True`                        |
| **issuperset(s2)**                  | Checks if set is superset of s2           | Check if s contains all elements | `{1,2,3}.issuperset({1})` → `True`                        |
| **pop()**                           | Removes and returns random element        | Remove unknown element           | `{1,2,3}.pop()` → e.g. `1`                                |
| **remove(x)**                       | Removes `x`; raises error if absent       | Strict removal                   | `{1,2}.remove(2)` → `{1}`                                 |
| **symmetric_difference(s2)**        | Returns elements in either set, not both  | XOR set operation                | `{1,2}.symmetric_difference({2,3})` → `{1,3}`             |
| **symmetric_difference_update(s2)** | Updates set with symmetric difference     | In-place XOR                     | `s={1,2}; s.symmetric_difference_update({2,3})` → `{1,3}` |
| **union(s2,... )**                  | Returns all unique elements from all sets | Combine many sets                | `{1,2}.union({2,3},{3,4})` → `{1,2,3,4}`                  |
| **update(s2)**                      | Adds all elements of s2                   | In-place merge                   | `s={1,2}; s.update({2,3})` → `{1,2,3}`                    |


```python
set1= {1,2}
set1.add(3)
print(set1)
set1.clear()
print(set1)
set2= {1,4,6,9,3}
set1= set2.copy()
print(set1)
set3= {5,3,7,8,0}
set2.difference(set3)
set2.difference_update(set3)
print(set3)
set3.discard(0)
print(set3)
set4={4,5,6,7}
set4.intersection(set3)
set4.remove(5)
print(set4)
```

# Function
A Function is block of code which run only when we call, this is reusable block of code which we call whenever we need in our code.

| **Function Type**                                                     | **Description**                                                    | **Example**                                                                                             | **Common Application**                                                                               |
| --------------------------------------------------------------------- | ------------------------------------------------------------------ | ------------------------------------------------------------------------------------------------------- | ---------------------------------------------------------------------------------------------------- |
| **Positional Argument Function**                                      | Requires arguments to be passed in the correct order.              | `python\ndef add(a, b):\n    return a + b\nprint(add(3, 4))  # 7`                                       | Most common form of functions, used when argument order is clear and fixed.                          |
| **Keyword Argument Function**                                         | Arguments are passed using parameter names; order does not matter. | `python\ndef greet(name, age):\n    return f\"{name}, {age} years\"\nprint(greet(age=25, name='John'))` | Useful for readability, avoids mistakes from wrong order.                                            |
| **Default Argument Function**                                         | Parameters have default values used if no value is supplied.       | `python\ndef info(name, country=\"USA\"):\n    return name, country\nprint(info(\"Alex\"))`             | Used to create flexible functions with optional parameters.                                          |
| **Variable-length Argument Function — *args**                         | Accepts any number of **positional** arguments as a tuple.         | `python\ndef total(*nums):\n    return sum(nums)\nprint(total(1,2,3,4))`                                | Useful for summing, aggregating, or handling unknown numbers of inputs.                              |
| **Variable-length Argument Function — **kwargs**                      | Accepts any number of **keyword** arguments as a dictionary.       | `python\ndef display(**info):\n    return info\nprint(display(name=\"Tom\", age=30))`                   | Used when you need flexible keyword arguments (e.g., API data).                                      |
| **Combined args + kwargs Function**                                   | Accepts both positional and keyword arbitrary arguments.           | `python\ndef func(*args, **kwargs):\n    print(args, kwargs)\nfunc(1,2, name=\"Raj\")`                  | Used for building wrappers, decorators, and flexible APIs.                                           |
| **Keyword-Only Argument Function**                                    | Arguments after `*` must be passed by keyword.                     | `python\ndef book(title, *, price):\n    return title, price\nbook(\"Python\", price=200)`              | Ensures required clarity of arguments (e.g., configuration parameters).                              |
| **Positional-Only Argument Function**                                 | Arguments before `/` **must** be positional.                       | `python\ndef divide(a, b, /):\n    return a / b\ndivide(10, 2)`                                         | Used in internal APIs to avoid overriding variable names (e.g., built-in functions).                 |
| **Pass-through Function**                                             | A function that receives arguments but does nothing (`pass`).      | `python\ndef dummy(a, b):\n    pass`                                                                    | Used for placeholders during development; commonly used in interface or abstract method definitions. |



```python
#positional argument function: In positional arguments, values are assigned to parameters according to their position or order
def student(name, surname, city):
    print("Name:", name)
    print("Surname:", surname)
    print("City:", city)

student("Abhishek", "Shriramwar", "Pune")

#Keyword Argument Function: In keyword arguments, you pass values using the parameter names.The order does not matter.
def student(name, surname, city):
    print("Name:", name)
    print("Surname:", surname)
    print("City:", city)

student(surname="Shriramwar", city="Pune", name="Abhishek")

#Default Argument Function: A default argument is a parameter that already has a value in the function definition. If you don't pass a value for that parameter, Python automatically uses the default.
def student(name, country="India"):
    print("Name:", name)
    print("Country:", country)

student("Abhishek")
student("Rahul", "USA")

#Variable-Length Arguments
""" What if you don't know how many arguments the user will pass?
For example, you want to calculate the sum of 2, 3, 5 or 10 numbers.
You could create separate functions, but Python provides *args.
*args allows a function to accept any number of positional arguments. """

def total(*numbers):
    print(numbers)
    print(type(numbers))
    print(sum(numbers))

total(10, 20, 30, 40)

```

```python
#Variable-Length Keyword Arguments — **kwargs
""" Now imagine you don't know how many keyword arguments will be passed.
For example, an employee might have a name, company, role, salary, location, etc.
You can use **kwargs.
It collects keyword arguments into a dictionary. """

def employee_details(**info):
    print(info)
    print(type(info))

employee_details(
    name="Abhishek",
    company="PwC",
    role="DevOps Engineer"
)

#Combined *args and **kwargs
"""You can use both in the same function.
   This allows you to accept:
- Any number of positional arguments.
- Any number of keyword arguments."""

def details(*args, **kwargs):
    print("Positional arguments:", args)
    print("Keyword arguments:", kwargs)

details(10, 20, 30, name="Abhishek")
```

```python
#Keyword-Only Arguments
"""Sometimes you want to force users to pass certain arguments using their parameter names.
Python provides * for this.
Any parameters after * must be passed as keyword arguments."""

def book(title, *, price):
    print("Book:", title)
    print("Price:", price)

book("Python Programming", price=500)

#Positional-Only Arguments
"""This is the opposite of keyword-only arguments.
Python uses / to indicate that parameters before it must be passed positionally."""

def student(name, surname, /):
    print("Name:", name)
    print("surname:", surname)

student("Abhishek", "Shriramwar")

#Pass-through / Placeholder Function
"""This one is slightly different.
It isn't really an argument-passing category. It describes a function that has been declared but whose implementation is not ready.
Python provides pass as a placeholder statement."""

def login():
    pass

def register():
    pass

def logout():
    pass    #we are just creating structure of project here and implement the function later, its just a pass through function
```

Typically used in `map()`, `filter()`, `sorted()`, and `reduce()` functions. Useful in functional programming.                                     |

Uses the `yield` keyword to produce a value one at a time, pausing its state between yields.        | `python<br>def count_up_to(n):<br>    count = 1<br>    while count <= n:<br>        yield count<br>        count += 1<br>gen = count_up_to(5)<br>for i in gen:<br>    print(i) <br> gen_count=count_up_to(6) <br> print(next(gen_count))`                                                                                                                                                                                                                                 | Used for iterating over large datasets, lazy evaluation, or streams of data where you don't want to load everything into memory at once (e.g., reading large files).                          |

It can be used to iterate over data one item at a time.                           | `python<br>class MyIterator:<br>    def __init__(self, start, end):<br>        self.current = start<br>        self.end = end<br>    def __iter__(self):<br>        return self<br>    def __next__(self):<br>        if self.current >= self.end:<br>            raise StopIteration<br>        self.current += 1<br>        return self.current - 1<br>it = MyIterator(0, 3)<br>for i in it:<br>    print(i)` | Useful for creating custom iterable objects. Can be used in loops or in any scenario where you need to iterate through data step by step, such as database queries or custom data structures. |

```python
#Standard Function (Normal Function)
#A standard function is a regular function created using the def keyword.

def add(x, y):
    return x + y
result = add(2, 3)
print(result)

#Lambda Function
# A lambda function is a small anonymous function created using the lambda keyword.Anonymous means it doesn't require a function name when defined.

add = lambda x, y: x + y
print(add(2, 3))

#Generator Function
"""A generator function is a function that uses the yield keyword to produce values one at a time.
Unlike a normal function that typically returns a final result, a generator pauses its execution after each yield and remembers its state.""""Abhishek (Person 1) - Chrome.lnk"

def numbers():
    yield 10
    yield 20
    yield 30
gen = numbers()
print(next(gen))
print(next(gen))
print(next(gen))

#Iterator (Custom Iterator Object)
"""An iterator is technically not a function type. It is an object that allows you to access elements one at a time.
An iterator implements two special methods:
- __iter__() – returns the iterator object.
- __next__() – returns the next item.
When there are no more items, __next__() raises StopIteration."""

numbers = [10, 20, 30]
it = iter(numbers)
print(next(it))
print(next(it))
print(next(it))

#Recursive Function
"""A recursive function is a function that calls itself to solve a problem.
It needs two important components:
1. Base case – the condition that stops recursion.
2. Recursive case – the function calling itself with a smaller or simpler problem."""

def factorial(n):

    if n == 0:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
```

## Exception Handling

try - contain code that may generate error

except - handle error

else - execute when no error occur

finally -always execute if error occurs or not

raise - manually raise exception

```python
try:
    num1 = 10
    num2 = 0

    if num2 < 0:
        raise ValueError("Negative divisor is not allowed")

    result = num1 / num2

except ValueError as e:
    print("Invalid input:", e)

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Division result:", result)

finally:
    print("Execution completed")
```
